```python

# Convert TotalCharges to numeric without removing decimal values
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"].astype(str).str.strip().replace("", pd.NA),
    errors="coerce"
)

# Check the result
print("Missing TotalCharges:", df["TotalCharges"].isna().sum())
print(df["TotalCharges"].head(10))
```

```python
df['TotalCharges']=df['TotalCharges'].fillna(0)
```

```python
df=df.drop(columns=['customerID'],errors='ignore')
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
x = pd.get_dummies(df.drop(columns=['Churn']),drop_first=True,dtype=int)
y=df['Churn']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
x
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=4000)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model.fit(x_train_scaled,y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_prob = model.predict_proba(x_test_scaled)[:, 1]
```

```python
from sklearn.metrics import precision_score, recall_score,accuracy_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print("Precision:", precision)
print("Recall:", recall)
```

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", roc_auc)
```

```python
from sklearn.metrics import average_precision_score

pr_auc = average_precision_score(y_test, y_prob)

print("PR-AUC:", pr_auc)
```

```python
print("Logistic Regression Results")
print("---------------------------")
print("Precision:", precision)
print("Recall:", recall)
print("ROC-AUC:", roc_auc)
print("PR-AUC:", pr_auc)
```

Initial Xgboost Experiment

```python
from xgboost import XGBClassifier
```

```python
model_xg=XGBClassifier(n_estimators=300,max_depth=4,learning_rate=0.5,scale_pos_weight=(y_train==0).sum()/(y_train==1).sum())
model_xg.fit(x_train,y_train)
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold

param_grid = {
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.05, 0.1],
    'n_estimators': [100, 200, 300]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

grid_search = GridSearchCV(
    estimator=XGBClassifier(
        scale_pos_weight=scale_pos_weight,
        eval_metric='logloss',
        random_state=42,
        n_jobs=1
    ),
    param_grid=param_grid,
    scoring='average_precision',
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(x_train, y_train)

best_model = grid_search.best_estimator_

print(grid_search.best_params_)
print(grid_search.best_score_)
```

```python
y_pred_xgb = best_model.predict(x_test)

y_prob_xgb = best_model.predict_proba(
    x_test
)[:, 1]
```

```python
print("Precision:", precision_score(y_test, y_pred_xgb))
print("Recall:", recall_score(y_test, y_pred_xgb))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_xgb))
print("PR-AUC:", average_precision_score(y_test, y_prob_xgb))
```

```python
from sklearn.model_selection import cross_val_predict

oof_prob = cross_val_predict(
    best_model,
    x_train,
    y_train,
    cv=cv,
    method='predict_proba',
    n_jobs=-1
)[:, 1]
```

```python
from sklearn.metrics import precision_recall_curve
import numpy as np

precision_values, recall_values, thresholds = precision_recall_curve(
    y_train,
    oof_prob
)

valid = np.where(recall_values[:-1] >= 0.80)[0]

if len(valid):
    best_idx = valid[np.argmax(precision_values[:-1][valid])]
    threshold = thresholds[best_idx]
    print("Selected threshold:", threshold)
```

```python
y_pred_tuned = (y_prob_xgb >= threshold).astype(int)

print("Precision:", precision_score(y_test, y_pred_tuned))
print("Recall:", recall_score(y_test, y_pred_tuned))
```

## Business Recommendations

Based on exploratory data analysis and SHAP model interpretation, the following strategies are recommended to reduce customer churn:

1. **Improve New Customer Retention:** Customers with shorter tenure show higher churn risk. Introduce onboarding programs, regular follow-ups, and early-stage engagement offers.

2. **Encourage Long-Term Contracts:** Customers with month-to-month contracts show higher churn risk. Provide incentives and exclusive benefits for customers switching to one-year or two-year contracts.

3. **Investigate Fiber Optic Services:** Fiber optic service is associated with increased churn predictions. Review service quality, pricing, and customer satisfaction to identify potential issues.

4. **Promote Automatic Payments:** Customers using electronic checks show higher churn predictions. Encourage convenient automatic payment options through suitable incentives.

5. **Review Pricing Strategies:** Higher monthly charges are associated with increased churn predictions. Introduce flexible plans and personalized discounts to improve affordability.

6. **Promote Additional Support Services:** Online security and technical support are associated with lower churn predictions. Promote these services to improve customer experience and retention.

**Overall Recommendation:** Prioritize early-tenure customers and month-to-month subscribers for retention campaigns, while using the XGBoost model to identify high-risk customers and allocate retention resources effectively.
